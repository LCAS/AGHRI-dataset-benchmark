# AGHRI 2D tracking: paper tables and figures

The overall comparisons use all six held-out recordings and both camera
streams. The three-recording plots illustrate the same footpath, inside
strawberry polytunnel, and outside vineyard settings used in the paper's
earlier figures. Detector-input plots use the validation-selected YOLOv8s
checkpoints; the ground-truth-box control is shown separately.

Run this notebook from `2d-tracking` or its `notebooks` directory. It reads
saved TrackEval results and tracker timings, so no models are rerun.

```python
from __future__ import annotations

import csv
import os
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

try:
    display
except NameError:
    display = print


def find_tracking_root():
    override = os.environ.get("AGHRI_TRACK_ROOT")
    if override:
        return Path(override).resolve()
    for candidate in (Path.cwd(), *Path.cwd().parents):
        if (candidate / "reports/aghri_test/trackeval_summary_by_source.csv").exists():
            return candidate
    raise FileNotFoundError("Run from the 2d-tracking repository or set AGHRI_TRACK_ROOT")


ROOT = find_tracking_root()
DATA = ROOT / "reports/aghri_test"
OUT = Path(os.environ.get("AGHRI_TRACK_PAPER_OUT", DATA / "paper_results"))
FIGURES = OUT / "figures"
FIGURES.mkdir(parents=True, exist_ok=True)

TRACKERS = [
    ("boxmot_bytetrack", "ByteTrack"),
    ("boxmot_ocsort", "OC-SORT"),
    ("boxmot_deepocsort", "Deep OC-SORT"),
    ("boxmot_botsort", "BoT-SORT"),
    ("boxmot_boosttrack", "BoostTrack"),
    ("boxmot_strongsort", "StrongSORT"),
    ("deepsort", "DeepSORT"),
    ("norfair", "Norfair"),
]
TRACKER_NAMES = dict(TRACKERS)
TRACKER_ORDER = [key for key, _ in TRACKERS]
DETECTOR_SOURCES = ("zed_yolov8s_pretrained", "fisheye_yolov8s_pretrained")
SELECTED = [
    ("footpath1_p1_oj+mk+gl_1walk+check_st_11_12_2024_1_label", "Footpath"),
    ("in_straw_3pick_diff_st_10_24_2024_5_a_label", "Inside strawberry\npolytunnel"),
    ("out_vine_4swap+walk_st_ly_11_06_2024_2_label", "Outside vineyard"),
]
SCENARIO_NAMES = dict(SELECTED)

plt.rcParams.update({
    "font.family": "DejaVu Sans", "font.size": 10,
    "axes.titlesize": 13, "axes.titleweight": "bold", "axes.labelsize": 11,
    "xtick.labelsize": 9, "ytick.labelsize": 10,
    "figure.facecolor": "white", "axes.spines.top": False,
    "axes.spines.right": False, "axes.grid": False,
    "pdf.fonttype": 42, "ps.fonttype": 42,
})


def save_figure(fig, name):
    fig.savefig(FIGURES / f"{name}.png", dpi=300, bbox_inches="tight")
    fig.savefig(FIGURES / f"{name}.pdf", bbox_inches="tight")
    plt.close(fig)


def save_csv(frame, name):
    frame.to_csv(OUT / name, index=False, float_format="%.10f")
```

## Load the saved test results

The paper tables use an equal mean over the 12 camera-specific streams for
rate metrics. Identity switches are summed. Tracker processing speed is the
equal mean of the 12 per-run frame rates. TrackEval MOTP is matched-box IoU,
where higher values are better.

```python
records = []
for source in (*DETECTOR_SOURCES, "gt"):
    for tracker in TRACKER_ORDER:
        path = DATA / "trackeval_detailed" / source / f"{tracker}.csv"
        with path.open(newline="", encoding="utf-8") as handle:
            for row in csv.DictReader(handle):
                if row["seq"] == "COMBINED":
                    continue
                stream = row["seq"]
                if stream.endswith("_zed_rgb"):
                    sequence = stream[:-len("_zed_rgb")]
                    camera = "ZED RGB"
                elif stream.endswith("_front_fisheye"):
                    sequence = stream[:-len("_front_fisheye")]
                    camera = "Front fisheye"
                else:
                    raise ValueError(stream)
                records.append({
                    "input": "Ground-truth-based" if source == "gt" else "Detector-based",
                    "source": source, "tracker": tracker,
                    "sequence": sequence, "camera": camera,
                    "mota": float(row["MOTA"]),
                    "motp": float(row["MOTP"]),
                    "idf1": float(row["IDF1"]),
                    "idp": float(row["IDP"]),
                    "idr": float(row["IDR"]),
                    "precision": float(row["CLR_Pr"]),
                    "recall": float(row["CLR_Re"]),
                    "idsw": int(float(row["IDSW"])),
                    "hota": float(row["HOTA___AUC"]),
                    "evaluated_frames": int(float(row["CLR_Frames"])),
                })
results = pd.DataFrame(records)
assert len(results) == 192
assert results.groupby(["input", "tracker"]).size().eq(12).all()

timing = pd.read_csv(DATA / "tracker_throughput_per_run.csv")
timing = timing[timing["source"].isin(DETECTOR_SOURCES)].copy()
assert len(timing) == 96
assert timing.groupby("tracker").size().eq(12).all()
```

## Overall detector-based tracking table

```python
detector = results[results["input"] == "Detector-based"].copy()
control = results[results["input"] == "Ground-truth-based"].copy()
overall = detector.groupby("tracker", sort=False).agg(
    mota=("mota", "mean"), motp=("motp", "mean"),
    idf1=("idf1", "mean"), idp=("idp", "mean"),
    idr=("idr", "mean"), idsw=("idsw", "sum"),
    hota=("hota", "mean"), camera_streams=("sequence", "size"),
)
overall["tracker_fps"] = timing.groupby("tracker")["tracking_fps"].mean()
overall = overall.reindex(TRACKER_ORDER).reset_index()
overall.insert(1, "tracker_name", overall["tracker"].map(TRACKER_NAMES))
assert overall["camera_streams"].eq(12).all()
save_csv(overall, "overall_detector_based.csv")
display(overall[["tracker_name", "mota", "motp", "idf1", "idp", "idr", "idsw", "tracker_fps"]])
```

## Detector-input versus ground-truth-box table

```python
comparison = results.groupby("input", sort=False).agg(
    mota=("mota", "mean"), idf1=("idf1", "mean"),
    precision=("precision", "mean"), recall=("recall", "mean"),
    idsw=("idsw", "sum"), tracker_camera_runs=("sequence", "size"),
)
comparison = comparison.reindex(["Detector-based", "Ground-truth-based"])
assert comparison["tracker_camera_runs"].eq(96).all()
delta = comparison.loc["Ground-truth-based"] - comparison.loc["Detector-based"]
comparison.loc["Difference (GT - detector)"] = delta
comparison = comparison.reset_index()
save_csv(comparison, "detector_vs_gt_overall.csv")
display(comparison)
```

## Three illustrative recordings

These are the same three recordings represented in the earlier paper figures.
Scenario-specific results are displayed separately from the six-recording
overall results. Participant counts come from the released sequence summary;
camera frame counts come from the annotation manifest.

```python
composition_path = Path(os.environ.get(
    "AGHRI_TRACK_COMPOSITION_CSV", DATA / "scenario_composition.csv"))
composition = pd.read_csv(composition_path)
selected_ids = [sequence for sequence, _ in SELECTED]
composition = composition.set_index("sequence").loc[selected_ids].reset_index()
assert composition["participants_in_sequence"].tolist() == [1, 3, 4]
save_csv(composition, "selected_scenario_composition.csv")

selected_results = results[results["sequence"].isin(selected_ids)].copy()
scenario_camera = selected_results.groupby(["sequence", "camera", "input"], sort=False).agg(
    mota=("mota", "mean"), idf1=("idf1", "mean"),
    idsw=("idsw", "sum"), tracker_runs=("tracker", "size"),
).reset_index()
assert scenario_camera["tracker_runs"].eq(8).all()
save_csv(scenario_camera, "selected_scenario_camera_results.csv")

scenario_combined = selected_results.groupby(["sequence", "input"], sort=False).agg(
    mota=("mota", "mean"), idf1=("idf1", "mean"),
    idsw=("idsw", "sum"), tracker_camera_runs=("tracker", "size"),
).reset_index()
assert scenario_combined["tracker_camera_runs"].eq(16).all()
save_csv(scenario_combined, "selected_scenario_combined_results.csv")
display(composition)
display(scenario_combined)
```

## Figure 1. Key tracker metrics

```python
bar_colors = ["#4B8FB7", "#306883", "#829CA5", "#B84E3D",
              "#CF8245", "#9B6D60", "#A68E77", "#667175"]
fig, axes = plt.subplots(1, 3, figsize=(17, 4.7))
x = np.arange(len(overall))
for ax, metric, title, direction, fmt in zip(
    axes, ("mota", "idf1", "idsw"),
    ("(a) MOTA", "(b) IDF1", "(c) Total IDSW"),
    ("higher is better", "higher is better", "lower is better"),
    (".4f", ".4f", ".0f"),
):
    values = overall[metric].to_numpy()
    bars = ax.bar(x, values, color=bar_colors, width=0.72)
    best_value = values.min() if metric == "idsw" else values.max()
    best_indices = set(np.flatnonzero(np.isclose(values, best_value)).tolist())
    for i in best_indices:
        bars[i].set_edgecolor("#222222")
        bars[i].set_linewidth(1.7)
    ax.set_title(title, pad=14)
    ax.set_ylabel(metric.upper() if metric == "idsw" else metric.upper())
    ax.set_xticks(x, overall["tracker_name"], rotation=42, ha="right")
    ax.set_ylim(0, max(1.05, values.max() * 1.17) if metric != "idsw" else values.max() * 1.18)
    ax.yaxis.grid(True, alpha=0.22)
    ax.set_axisbelow(True)
    ax.text(0.99, 0.98, direction, transform=ax.transAxes,
            ha="right", va="top", fontsize=8, color="#777777")
    for i, (bar, value) in enumerate(zip(bars, values)):
        ax.annotate(format(value, fmt),
                    (bar.get_x() + bar.get_width() / 2, bar.get_height()),
                    xytext=(0, 3), textcoords="offset points", ha="center",
                    va="bottom", fontsize=8, fontweight="bold" if i in best_indices else "normal")
fig.tight_layout(w_pad=2.2)
save_figure(fig, "aghri_2d_tracking_key_metrics")
```

## Figure 2. Composition of the three selected recordings

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 4.8))
scenario_labels = [label for _, label in SELECTED]
x = np.arange(3)
w = 0.31
fisheye_bars = axes[0].bar(x - w / 2, composition["front_fisheye_frames"], w,
                           color="#5C6BC0", label="Front fisheye")
zed_bars = axes[0].bar(x + w / 2, composition["zed_rgb_frames"], w,
                       color="#26A69A", label="ZED RGB")
axes[0].set_title("(a) Annotation-backed frames per recording and camera")
axes[0].set_ylabel("Frame count")
axes[0].set_xticks(x, scenario_labels)
axes[0].legend(loc="upper left", fontsize=9)
for bars in (fisheye_bars, zed_bars):
    axes[0].bar_label(bars, padding=3, fontsize=9)

participant_bars = axes[1].bar(x, composition["participants_in_sequence"],
                               color=["#4CAF50", "#FF9800", "#F44336"], width=0.53)
axes[1].set_title("(b) Participants per recording")
axes[1].set_ylabel("Number of participants")
axes[1].set_xticks(x, scenario_labels)
axes[1].set_ylim(0, 4.7)
axes[1].bar_label(participant_bars, padding=3, fontsize=10, fontweight="bold")
for ax in axes:
    ax.yaxis.grid(True, alpha=0.22)
    ax.set_axisbelow(True)
fig.tight_layout(w_pad=2.7)
save_figure(fig, "aghri_2d_tracking_scenario_composition")
```

## Figure 3. Detector input versus ground-truth-box input

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5.1))
for ax, metric, title in zip(axes, ("mota", "idf1"), ("(a) MOTA", "(b) IDF1")):
    for i, (sequence, _) in enumerate(SELECTED):
        rows = scenario_combined[scenario_combined["sequence"] == sequence].set_index("input")
        det = rows.loc["Detector-based", metric]
        gt = rows.loc["Ground-truth-based", metric]
        ax.plot([i - 0.055, i + 0.055], [det, gt], color="#999999", lw=1.8, zorder=1)
        ax.scatter(i - 0.055, det, s=72, color="#245278", zorder=3,
                   label="Detector-based" if i == 0 else None)
        ax.scatter(i + 0.055, gt, s=72, color="#D77D27", marker="s", zorder=3,
                   label="Ground-truth-based" if i == 0 else None)
        ax.annotate(f"{det:.4f}", (i - 0.055, det), xytext=(-5, -17),
                    textcoords="offset points", ha="right", fontsize=8,
                    color="#245278", fontweight="bold")
        ax.annotate(f"{gt:.4f}", (i + 0.055, gt), xytext=(5, 8),
                    textcoords="offset points", ha="left", fontsize=8,
                    color="#D77D27", fontweight="bold")
        difference_pp = (gt - det) * 100
        difference_label = "0.00 pp" if abs(difference_pp) < 1e-8 else f"{difference_pp:+.2f} pp"
        ax.annotate(difference_label, (i, max(det, gt)),
                    xytext=(0, 25), textcoords="offset points", ha="center",
                    fontsize=8, color="#777777", fontweight="bold")
    ax.set_title(title, pad=23)
    ax.set_ylabel(metric.upper())
    ax.set_xticks(np.arange(3), scenario_labels)
    values = scenario_combined[metric]
    low = max(0, values.min() - 0.11)
    ax.set_ylim(low, 1.06)
    ax.yaxis.grid(True, alpha=0.22)
    ax.set_axisbelow(True)
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, loc="upper center", ncol=2, frameon=True,
           bbox_to_anchor=(0.5, 1.05))
fig.tight_layout(rect=(0, 0, 1, 0.94), w_pad=2.0)
save_figure(fig, "aghri_2d_tracking_detector_vs_gt_combined_cameras_scenario_lines")
```

## Figure 4. Camera-specific detector-input tracking

```python
fig, axes = plt.subplots(1, 2, figsize=(15.8, 5.0))
for ax, metric, title in zip(axes, ("mota", "idf1"), ("(a) MOTA", "(b) IDF1")):
    plot_rows = scenario_camera[scenario_camera["input"] == "Detector-based"]
    fish = [plot_rows[(plot_rows["sequence"] == seq) &
                      (plot_rows["camera"] == "Front fisheye")][metric].iloc[0]
            for seq, _ in SELECTED]
    zed = [plot_rows[(plot_rows["sequence"] == seq) &
                     (plot_rows["camera"] == "ZED RGB")][metric].iloc[0]
           for seq, _ in SELECTED]
    bars_a = ax.bar(x - 0.18, fish, 0.36, color="#5C6BC0", label="Front fisheye")
    bars_b = ax.bar(x + 0.18, zed, 0.36, color="#26A69A", label="ZED RGB")
    ax.set_title(title, pad=14)
    ax.set_ylabel(metric.upper())
    ax.set_xticks(x, scenario_labels)
    ax.set_ylim(0, 1.09)
    ax.yaxis.grid(True, alpha=0.22)
    ax.set_axisbelow(True)
    for bars in (bars_a, bars_b):
        ax.bar_label(bars, fmt="%.4f", padding=3, fontsize=8)
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, loc="upper center", ncol=2, frameon=True,
           bbox_to_anchor=(0.5, 1.04))
fig.tight_layout(rect=(0, 0, 1, 0.94), w_pad=2.0)
save_figure(fig, "aghri_2d_tracking_camera_comparison_by_scenario")
```

## Paper-ready LaTeX tables

```python
lines = [
    r"\begin{table*}[t!]", r"\centering",
    r"\caption{Overall detector-based 2D tracking results across 12 camera-specific streams from six AGHRI test recordings. Values for rate metrics are equal-stream means; IDSW is summed.}",
    r"\label{tab:2d_tracking_overall_results}", r"\scriptsize",
    r"\setlength{\tabcolsep}{4pt}", r"\renewcommand{\arraystretch}{1.15}",
    r"\begin{tabular}{|l|c|c|c|c|c|c|c|}", r"\hline",
    r"Tracker & MOTA $\uparrow$ & MOTP $\uparrow$ & IDF1 $\uparrow$ & IDP $\uparrow$ & IDR $\uparrow$ & IDSW $\downarrow$ & Tracker frames/s $\uparrow$ " + "\\\\",
    r"\hline",
]
for row in overall.itertuples():
    lines.append(f"{row.tracker_name} & {row.mota:.4f} & {row.motp:.4f} & {row.idf1:.4f} & {row.idp:.4f} & {row.idr:.4f} & {row.idsw:d} & {row.tracker_fps:.2f} " + r"\\")
    lines.append(r"\hline")
lines += [r"\end{tabular}", r"\end{table*}"]
(OUT / "overall_detector_based.tex").write_text("\n".join(lines) + "\n", encoding="utf-8")

lines = [
    r"\begin{table*}[t!]", r"\centering",
    r"\caption{Comparison of detector-based and ground-truth-box tracking across 12 camera-specific streams and eight trackers. Rate metrics are equal-run means; IDSW is summed.}",
    r"\label{tab:2d_tracking_detection_input_comparison}", r"\scriptsize",
    r"\setlength{\tabcolsep}{5pt}", r"\renewcommand{\arraystretch}{1.15}",
    r"\begin{tabular}{|l|c|c|c|c|c|}", r"\hline",
    r"Input setting & MOTA $\uparrow$ & IDF1 $\uparrow$ & Precision $\uparrow$ & Recall $\uparrow$ & Total IDSW $\downarrow$ " + "\\\\",
    r"\hline",
]
for row in comparison.itertuples(index=False):
    is_difference = row.input == "Difference (GT - detector)"
    label = "Difference (GT $-$ detector)" if is_difference else row.input
    format_rate = "+.4f" if is_difference else ".4f"
    format_count = "+.0f" if is_difference else ".0f"
    lines.append(
        f"{label} & {row.mota:{format_rate}} & {row.idf1:{format_rate}} & "
        f"{row.precision:{format_rate}} & {row.recall:{format_rate}} & "
        f"{row.idsw:{format_count}} " + r"\\")
    lines.append(r"\hline")
lines += [r"\end{tabular}", r"\end{table*}"]
(OUT / "detector_vs_gt_overall.tex").write_text("\n".join(lines) + "\n", encoding="utf-8")

print(f"Wrote two tables and four figures to {OUT}")
```
